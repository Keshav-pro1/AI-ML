```python
text="Name the Largest animal on land"
token_generated = tiktoken.encoding_for_model(model_name="gpt-4")
tokenId = token_generated.encode(text)
tokenId
```

```python
decodeId = [678, 279, 84419, 10065, 389, 4363]
```

```python
result = token_generated.decode(decodeId)
result
```
